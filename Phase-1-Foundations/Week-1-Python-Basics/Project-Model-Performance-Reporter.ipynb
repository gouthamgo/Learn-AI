## 🏗️ Part 1: Core Functions

Let's build the foundation with reusable functions.

```python
def calculate_accuracy(correct_predictions, total_predictions):
    """
    Calculate model accuracy percentage.
    
    Args:
        correct_predictions (int): Number of correct predictions
        total_predictions (int): Total number of predictions
    
    Returns:
        float: Accuracy as percentage (0-100)
    """
    if total_predictions == 0:
        return 0.0
    return (correct_predictions / total_predictions) * 100


def calculate_precision(true_positives, false_positives):
    """
    Calculate precision: How many predicted positives were actually correct?
    """
    total = true_positives + false_positives
    if total == 0:
        return 0.0
    return (true_positives / total) * 100


def calculate_recall(true_positives, false_negatives):
    """
    Calculate recall: How many actual positives did we find?
    """
    total = true_positives + false_negatives
    if total == 0:
        return 0.0
    return (true_positives / total) * 100


def calculate_f1_score(precision, recall):
    """
    Calculate F1 Score: Harmonic mean of precision and recall
    """
    if precision + recall == 0:
        return 0.0
    return 2 * (precision * recall) / (precision + recall)


# Test the functions
print("Testing Core Functions:")
print("=" * 50)
print(f"Accuracy: {calculate_accuracy(85, 100):.2f}%")
print(f"Precision: {calculate_precision(80, 20):.2f}%")
print(f"Recall: {calculate_recall(80, 15):.2f}%")
print(f"F1 Score: {calculate_f1_score(80, 84.21):.2f}")
```

## 📊 Part 2: Model Evaluation Class

Create a complete model tracking system.

```python
def evaluate_model_performance(model_name, metrics):
    """
    Evaluate overall model performance and assign grade.
    
    Args:
        model_name (str): Name of the model
        metrics (dict): Dictionary with accuracy, precision, recall, f1_score
    
    Returns:
        dict: Enhanced metrics with grade and recommendations
    """
    accuracy = metrics['accuracy']
    
    # Determine grade and status
    if accuracy >= 95:
        grade = 'A+'
        status = 'Excellent'
        recommendation = 'Ready for production deployment'
    elif accuracy >= 90:
        grade = 'A'
        status = 'Very Good'
        recommendation = 'Minor tuning recommended'
    elif accuracy >= 85:
        grade = 'B+'
        status = 'Good'
        recommendation = 'Consider hyperparameter tuning'
    elif accuracy >= 80:
        grade = 'B'
        status = 'Acceptable'
        recommendation = 'Needs improvement - review features'
    elif accuracy >= 70:
        grade = 'C'
        status = 'Below Average'
        recommendation = 'Significant improvement needed'
    else:
        grade = 'F'
        status = 'Poor'
        recommendation = 'Consider different model architecture'
    
    return {
        'model_name': model_name,
        'grade': grade,
        'status': status,
        'recommendation': recommendation,
        **metrics
    }


# Test with sample data
sample_metrics = {
    'accuracy': 92.5,
    'precision': 90.0,
    'recall': 88.5,
    'f1_score': 89.2
}

result = evaluate_model_performance("Image Classifier v1", sample_metrics)
print("\nModel Evaluation:")
print("=" * 50)
for key, value in result.items():
    print(f"{key.replace('_', ' ').title()}: {value}")
```

## 📝 Part 3: Report Generation

Create professional, formatted reports.

```python
def generate_text_report(evaluation_result):
    """
    Generate a professional text report.
    """
    report = f"""
{'='*60}
           AI MODEL PERFORMANCE REPORT
{'='*60}

Model Name: {evaluation_result['model_name']}
Overall Grade: {evaluation_result['grade']}
Status: {evaluation_result['status']}

{'='*60}
PERFORMANCE METRICS
{'='*60}

Accuracy:  {evaluation_result['accuracy']:.2f}%
Precision: {evaluation_result['precision']:.2f}%
Recall:    {evaluation_result['recall']:.2f}%
F1 Score:  {evaluation_result['f1_score']:.2f}

{'='*60}
RECOMMENDATION
{'='*60}

{evaluation_result['recommendation']}

{'='*60}
"""
    return report


# Generate and display report
report = generate_text_report(result)
print(report)
```

## 🔄 Part 4: Multi-Model Comparison

Compare multiple models side-by-side.

```python
def compare_models(models_data):
    """
    Compare multiple models and identify best performer.
    
    Args:
        models_data (list): List of tuples (model_name, metrics_dict)
    
    Returns:
        dict: Comparison results
    """
    evaluated_models = []
    
    for model_name, metrics in models_data:
        eval_result = evaluate_model_performance(model_name, metrics)
        evaluated_models.append(eval_result)
    
    # Find best model by accuracy
    best_model = max(evaluated_models, key=lambda x: x['accuracy'])
    
    return {
        'all_models': evaluated_models,
        'best_model': best_model,
        'total_models': len(evaluated_models)
    }


def print_comparison_report(comparison_result):
    """
    Print a formatted comparison report.
    """
    print("\n" + "="*70)
    print("           MULTI-MODEL COMPARISON REPORT")
    print("="*70)
    print(f"\nTotal Models Evaluated: {comparison_result['total_models']}")
    print("\n" + "-"*70)
    print(f"{'Model Name':<30} {'Accuracy':<12} {'Grade':<8} {'Status':<15}")
    print("-"*70)
    
    for model in comparison_result['all_models']:
        print(f"{model['model_name']:<30} {model['accuracy']:>6.2f}%     {model['grade']:<8} {model['status']:<15}")
    
    print("="*70)
    best = comparison_result['best_model']
    print(f"\n🏆 BEST MODEL: {best['model_name']}")
    print(f"   Accuracy: {best['accuracy']:.2f}%")
    print(f"   Grade: {best['grade']}")
    print(f"   Recommendation: {best['recommendation']}")
    print("\n" + "="*70)


# Test with multiple models
models_to_compare = [
    ("Image Classifier v1", {'accuracy': 92.5, 'precision': 90.0, 'recall': 88.5, 'f1_score': 89.2}),
    ("Spam Detector v2", {'accuracy': 87.3, 'precision': 85.5, 'recall': 83.0, 'f1_score': 84.2}),
    ("Sentiment Analyzer v1", {'accuracy': 78.8, 'precision': 76.0, 'recall': 75.5, 'f1_score': 75.7}),
    ("Fraud Detection v3", {'accuracy': 96.2, 'precision': 95.8, 'recall': 94.5, 'f1_score': 95.1}),
    ("Chatbot Intent Classifier", {'accuracy': 89.5, 'precision': 88.0, 'recall': 87.2, 'f1_score': 87.6})
]

comparison = compare_models(models_to_compare)
print_comparison_report(comparison)
```

## 💾 Part 5: Export Functionality

Save reports to files for sharing with stakeholders.

```python
def save_report_to_file(report_text, filename=f"{OUT}/model_report.txt"):
    """
    Save report to a text file.
    
    Args:
        report_text (str): The report content
        filename (str): Output filename
    """
    try:
        with open(filename, 'w') as f:
            f.write(report_text)
        print(f"✅ Report saved successfully to: {filename}")
        return True
    except Exception as e:
        print(f"❌ Error saving report: {e}")
        return False


def export_comparison_csv(comparison_result, filename=f"{OUT}/model_comparison.csv"):
    """
    Export comparison data to CSV format.
    """
    try:
        with open(filename, 'w') as f:
            # Write header
            f.write("Model Name,Accuracy,Precision,Recall,F1 Score,Grade,Status,Recommendation\n")
            
            # Write data
            for model in comparison_result['all_models']:
                f.write(f"{model['model_name']},")
                f.write(f"{model['accuracy']:.2f},")
                f.write(f"{model['precision']:.2f},")
                f.write(f"{model['recall']:.2f},")
                f.write(f"{model['f1_score']:.2f},")
                f.write(f"{model['grade']},")
                f.write(f"{model['status']},")
                f.write(f"{model['recommendation']}\n")
        
        print(f"✅ CSV exported successfully to: {filename}")
        return True
    except Exception as e:
        print(f"❌ Error exporting CSV: {e}")
        return False


# Test export functions
print("\nExporting Reports...")
print("="*50)

# Save individual report
individual_report = generate_text_report(result)
save_report_to_file(individual_report, f"{OUT}/sample_model_report.txt")

# Export comparison CSV
export_comparison_csv(comparison, f"{OUT}/model_comparison.csv")
```

## 🎯 Part 6: Complete System Integration

Put it all together in a user-friendly interface.

```python
class ModelPerformanceReporter:
    """
    Complete model performance reporting system.
    """
    
    def __init__(self):
        self.models = []
    
    def add_model(self, model_name, correct, total, true_pos, false_pos, false_neg):
        """
        Add a model with its performance data.
        """
        accuracy = calculate_accuracy(correct, total)
        precision = calculate_precision(true_pos, false_pos)
        recall = calculate_recall(true_pos, false_neg)
        f1 = calculate_f1_score(precision, recall)
        
        metrics = {
            'accuracy': accuracy,
            'precision': precision,
            'recall': recall,
            'f1_score': f1
        }
        
        self.models.append((model_name, metrics))
        print(f"✅ Added model: {model_name}")
    
    def generate_full_report(self):
        """
        Generate comprehensive report for all models.
        """
        if not self.models:
            print("❌ No models to report on. Add models first.")
            return
        
        comparison = compare_models(self.models)
        print_comparison_report(comparison)
        return comparison
    
    def export_reports(self, base_filename="model_report"):
        """
        Export all reports to files.
        """
        comparison = compare_models(self.models)
        export_comparison_csv(comparison, f"{base_filename}.csv")
        
        # Save detailed report for best model
        best = comparison['best_model']
        best_report = generate_text_report(best)
        save_report_to_file(best_report, f"{base_filename}_best_model.txt")


# Demo: Use the complete system
print("\n" + "="*70)
print("         DEMONSTRATION: COMPLETE REPORTING SYSTEM")
print("="*70)

reporter = ModelPerformanceReporter()

# Add models with raw data
print("\nAdding models...")
reporter.add_model("ResNet50 Image Classifier", 925, 1000, 880, 95, 25)
reporter.add_model("BERT Sentiment Analyzer", 873, 1000, 820, 120, 60)
reporter.add_model("XGBoost Fraud Detector", 962, 1000, 940, 40, 20)
reporter.add_model("LSTM Stock Predictor", 788, 1000, 750, 180, 70)

# Generate full report
reporter.generate_full_report()

# Export everything
print("\n" + "-"*70)
print("Exporting reports...")
reporter.export_reports("final_model_analysis")
```
